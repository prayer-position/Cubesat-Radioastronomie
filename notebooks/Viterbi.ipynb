```python
import sys
import os 

sys.path.append(os.path.abspath(os.path.join('..')))

from src.encoders import viterbi
from src.modulation import gmsk_mod, gmsk_demod
from src.channel import awgn
import numpy as np
```

```python
# Global Parameters
BT = 0.3
Tb = 10
L = 16
k = 4

fc = 146e6
```

```python
bits = np.random.randint(0, 2, 10000)
```

```python
_, s_clean = gmsk_mod(np.random.randint(2, size=10000), fc, L, BT)
s_noisy = awgn(s_clean, 5)
clean_hat = gmsk_demod(s_clean, L)
noisy_hat = gmsk_demod(s_noisy, L)
```

```python
error_noisy = np.mean(noisy_hat != bits[:len(noisy_hat)])
error_clean = np.mean(clean_hat != bits[:len(clean_hat)])
```

```python
vit_clean = viterbi(clean_hat)
vit_noisy = viterbi(noisy_hat)
```

```python
error_vit_noisy = np.mean(vit_noisy != bits[:len(vit_noisy)])
error_vit_clean = np.mean(vit_clean != bits[:len(vit_clean)])
```

```python
print(f"Clean signal error : {error_clean}")
print(f"Noisy signal error : {error_noisy}")
print(f"Viterbi algorithm on clean signal error : {error_vit_clean}")
print(f"Viterbi algorithm on noisy signal error : {error_vit_noisy}")
```
